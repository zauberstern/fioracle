## 3. Understanding the Visualization

### **Row 1: Input Drivers (Top)**

Three key macro indicators, all normalized to standard deviations:

1. **Geopolitical Risk (GPR)** (Left)
   - Measures global conflict and tension
   - Spikes during 9/11, Iraq War, crises
   - Red shading = extreme risk periods

2. **Economic Policy Uncertainty (EPU)** (Middle)
   - Tracks policy-related economic uncertainty
   - Rises during elections, regulatory changes
   - Orange shading = high uncertainty

3. **Market Volatility (VIX)** (Right)
   - Market fear gauge
   - Spikes during market stress
   - Purple shading = high volatility

### **Row 2: Macro Regime Evolution (Middle)**

3-State HMM identifies market environment:
- **Blue**: Risk-Off (defensive conditions)
- **Purple**: Transitional (mixed signals)
- **Orange**: Risk-On (growth conditions)

Stacked area shows regime probabilities over time.

### **Row 3: Regime Distribution (Lower-Middle)**

Horizontal bars show bull vs bear time for each asset (2001-2010):
- **Green**: Bull regime %
- **Red**: Bear regime %

### **Row 4: Asset Regime Timeline (Bottom)**

Heatmap showing regime evolution:
- **Green cells**: Bull regime
- **Red cells**: Bear regime
- **Dashed box**: GFC period (2007-2009)

Notice how assets synchronize during crises!

## 6. Key Insights

From the visualizations, you should observe:

### **Crisis Patterns**
1. Macro shocks (GPR/EPU/VIX spikes) → Regime shifts (Risk-Off) → Asset regime changes (Bull→Bear)
2. Assets tend to synchronize during major crises (all red in heatmap)
3. Recovery shows gradual transition through intermediate regimes

### **Asset Behavior**
1. Different assets spend different amounts of time in bull/bear regimes
2. Some assets (e.g., gold) may be in bull regimes during risk-off periods
3. Diversification benefits visible in regime desynchronization

### **Predictive Power**
1. Macro indicators provide early warning signals
2. HMM captures regime transitions before full market moves
3. Asset-level models add granularity for portfolio construction
